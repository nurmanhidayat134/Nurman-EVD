```python
# Import library
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Mengatur tampilan grafik
sns.set_theme(style="whitegrid")

# Upload dataset
uploaded = files.upload()

# Membaca dataset
df = pd.read_csv('penjualan_nusantaramart.csv')

# Menampilkan 5 data pertama
print("5 Data Pertama:")
display(df.head())

# Informasi dataset
print("\nInformasi Dataset:")
df.info()

# Statistik deskriptif
print("\nStatistik Deskriptif:")
display(df.describe())

# Mengecek data kosong
print("\nJumlah Missing Value:")
print(df.isnull().sum())
```

```python
# Menghitung rata-rata total pembelian per kategori
rata_kategori = df.groupby('kategori_produk')['total'].mean()

print("Rata-rata Total Pembelian per Kategori:")
print(rata_kategori.sort_values(ascending=False))

# Mencari kategori dengan rata-rata tertinggi
kategori_tertinggi = rata_kategori.idxmax()
nilai_tertinggi = rata_kategori.max()

print("\nKategori dengan Rata-rata Tertinggi:")
print(kategori_tertinggi, "dengan rata-rata Rp", round(nilai_tertinggi, 2))

# Membandingkan mean dan median
mean_total = df['total'].mean()
median_total = df['total'].median()

print("\nPerbandingan Mean dan Median:")
print("Mean   :", round(mean_total, 2))
print("Median :", round(median_total, 2))

# Kesimpulan otomatis
if mean_total > median_total:
    print("Kesimpulan: Sebaran data cenderung menceng ke kanan (right-skewed).")
elif mean_total < median_total:
    print("Kesimpulan: Sebaran data cenderung menceng ke kiri (left-skewed).")
else:
    print("Kesimpulan: Mean dan median relatif sama, sehingga distribusi cenderung simetris.")
```

- Kategori produk dengan rata-rata nilai transaksi tertinggi adalah kategori yang memiliki nilai mean terbesar berdasarkan hasil pengelompokan data.
- Data kosong dapat diketahui melalui df.info() pada bagian Non-Null Count. Jika jumlah data non-null lebih sedikit daripada jumlah baris, berarti terdapat missing value. Pengecekan lebih rinci dapat menggunakan df.isnull().sum().
- Jika nilai mean lebih besar daripada median, distribusi data cenderung menceng ke kanan. Hal ini menunjukkan adanya beberapa transaksi bernilai tinggi yang memengaruhi rata-rata keseluruhan.

```python
# Menghitung total penjualan per kategori
total_kategori = df.groupby('kategori_produk')['total'].sum().sort_values(ascending=False)

# Membuat bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=total_kategori.values,
    y=total_kategori.index,
    palette='Blues_r'
)

plt.title('Total Penjualan per Kategori Produk')
plt.xlabel('Total Penjualan (Rp)')
plt.ylabel('Kategori Produk')
plt.tight_layout()
plt.show()

# Menampilkan kategori tertinggi dan terendah
print("Kategori Penjualan Tertinggi:")
print(total_kategori.idxmax(), "Rp", total_kategori.max())

print("\nKategori Penjualan Terendah:")
print(total_kategori.idxmin(), "Rp", total_kategori.min())
```

Kategori dengan kontribusi penjualan tertinggi adalah kategori yang memiliki total penjualan paling besar pada grafik.



Kategori dengan kontribusi penjualan terendah adalah kategori dengan total penjualan paling kecil.



Rekomendasi bisnis: Manajer Nusantara Mart dapat mempertahankan ketersediaan stok pada kategori dengan penjualan tertinggi. Sementara itu, kategori dengan penjualan rendah perlu dievaluasi melalui promosi, harga, dan minat pelanggan agar penjualannya meningkat.

```python
# Mengubah kolom tanggal menjadi format datetime
df['tanggal'] = pd.to_datetime(df['tanggal'], errors='coerce')

# Mengelompokkan penjualan berdasarkan bulan
df['bulan'] = df['tanggal'].dt.to_period('M')

tren_bulanan = df.groupby('bulan')['total'].sum().sort_index()

# Membuat line chart
plt.figure(figsize=(8, 5))

plt.plot(
    tren_bulanan.index.astype(str),
    tren_bulanan.values,
    marker='o',
    color='crimson'
)

plt.title('Tren Penjualan Bulanan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan (Rp)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Mencari bulan dengan penjualan tertinggi
bulan_tertinggi = tren_bulanan.idxmax()
penjualan_tertinggi = tren_bulanan.max()

print("Bulan dengan Penjualan Tertinggi:")
print(bulan_tertinggi, "dengan total Rp", penjualan_tertinggi)

# Analisis tren
if len(tren_bulanan) > 1:
    if tren_bulanan.iloc[-1] > tren_bulanan.iloc[0]:
        print("Kesimpulan: Penjualan secara keseluruhan cenderung meningkat.")
    elif tren_bulanan.iloc[-1] < tren_bulanan.iloc[0]:
        print("Kesimpulan: Penjualan secara keseluruhan cenderung menurun.")
    else:
        print("Kesimpulan: Penjualan awal dan akhir periode relatif sama.")
```

- Bulan dengan penjualan tertinggi dapat diketahui dari titik tertinggi pada grafik atau hasil bulan_tertinggi.
- Kemungkinan penyebab peningkatan penjualan antara lain adanya promo, diskon, hari besar, atau meningkatnya kebutuhan pelanggan pada periode tertentu. Namun, penyebab tersebut perlu diperiksa lebih lanjut karena tidak dapat dipastikan hanya dari grafik.
- Tren penjualan dapat mengalami kenaikan, penurunan, maupun fluktuasi. Jika grafik menunjukkan perubahan naik dan turun secara bergantian, berarti penjualan bersifat fluktuatif.

```python
# Membuat scatter plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='jumlah',
    y='total',
    hue='kategori_produk'
)

plt.title('Hubungan Jumlah Pembelian dengan Total Nilai Transaksi')
plt.xlabel('Jumlah Barang')
plt.ylabel('Total (Rp)')
plt.tight_layout()
plt.show()

# Menghitung korelasi
korelasi = df['jumlah'].corr(df['total'])

print("Nilai Korelasi:", round(korelasi, 3))

if korelasi > 0:
    print("Kesimpulan: Terdapat hubungan positif antara jumlah barang dan total transaksi.")
elif korelasi < 0:
    print("Kesimpulan: Terdapat hubungan negatif antara jumlah barang dan total transaksi.")
else:
    print("Kesimpulan: Tidak terdapat korelasi linear antara kedua variabel.")
```

Jika titik-titik pada grafik cenderung bergerak dari kiri bawah ke kanan atas, berarti terdapat hubungan positif antara jumlah barang dan total nilai transaksi. Semakin banyak barang yang dibeli, umumnya semakin besar nilai transaksinya.



Titik yang berada jauh dari kumpulan titik lainnya dapat menunjukkan kemungkinan outlier.



Kategori outlier dapat diketahui berdasarkan warna titik pada grafik. Transaksi tersebut perlu diperiksa lebih lanjut karena bisa disebabkan oleh pembelian dalam jumlah besar atau transaksi khusus.

```python
# Histogram distribusi rating pelanggan
plt.figure(figsize=(7, 5))

sns.histplot(
    df['rating_pelanggan'].dropna(),
    bins=5,
    color='seagreen'
)

plt.title('Distribusi Rating Pelanggan')
plt.xlabel('Rating')
plt.ylabel('Frekuensi')
plt.show()


# Boxplot jumlah pembelian per kategori
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x='kategori_produk',
    y='jumlah'
)

plt.title('Sebaran Jumlah Pembelian per Kategori Produk')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# Menghitung rata-rata rating
rata_rating = df['rating_pelanggan'].mean()

print("Rata-rata Rating Pelanggan:", round(rata_rating, 2))

if rata_rating >= 4:
    print("Kesimpulan: Rating pelanggan cenderung tinggi.")
elif rata_rating >= 3:
    print("Kesimpulan: Rating pelanggan berada pada tingkat sedang.")
else:
    print("Kesimpulan: Rating pelanggan cenderung rendah.")


# Mendeteksi outlier menggunakan metode IQR
print("\nKategori dengan Outlier:")

for kategori in df['kategori_produk'].dropna().unique():

    data = df[df['kategori_produk'] == kategori]['jumlah'].dropna()

    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    outlier = data[(data < batas_bawah) | (data > batas_atas)]

    if len(outlier) > 0:
        print(kategori, ":", len(outlier), "outlier")
    else:
        print(kategori, ": Tidak terdapat outlier")
```

Distribusi rating pelanggan dapat diamati melalui histogram. Jika sebagian besar data terkumpul pada nilai rating tinggi, berarti pelanggan cenderung memberikan penilaian positif.



Kategori produk yang memiliki titik di luar batas boxplot menunjukkan adanya kemungkinan outlier.



Outlier pada jumlah pembelian dapat disebabkan oleh transaksi dalam jumlah besar, pembelian untuk kebutuhan usaha, atau pembelian oleh reseller. Namun, penyebab sebenarnya perlu dikonfirmasi melalui pemeriksaan data transaksi.

```python
# Membuat dashboard dengan 4 grafik
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# Grafik 1: Total penjualan per kategori
sns.barplot(
    x=total_kategori.values,
    y=total_kategori.index,
    ax=axes[0, 0],
    palette='Blues_r'
)

axes[0, 0].set_title('Total Penjualan per Kategori')
axes[0, 0].set_xlabel('Total Penjualan (Rp)')
axes[0, 0].set_ylabel('Kategori Produk')


# Grafik 2: Tren penjualan bulanan
axes[0, 1].plot(
    tren_bulanan.index.astype(str),
    tren_bulanan.values,
    marker='o'
)

axes[0, 1].set_title('Tren Penjualan Bulanan')
axes[0, 1].set_xlabel('Bulan')
axes[0, 1].set_ylabel('Total Penjualan (Rp)')
axes[0, 1].tick_params(axis='x', rotation=45)


# Grafik 3: Distribusi rating pelanggan
sns.histplot(
    df['rating_pelanggan'].dropna(),
    bins=5,
    ax=axes[1, 0],
    color='seagreen'
)

axes[1, 0].set_title('Distribusi Rating Pelanggan')
axes[1, 0].set_xlabel('Rating')
axes[1, 0].set_ylabel('Frekuensi')


# Grafik 4: Sebaran jumlah pembelian
sns.boxplot(
    data=df,
    x='kategori_produk',
    y='jumlah',
    ax=axes[1, 1]
)

axes[1, 1].set_title('Sebaran Jumlah per Kategori')
axes[1, 1].set_xlabel('Kategori Produk')
axes[1, 1].set_ylabel('Jumlah')
axes[1, 1].tick_params(axis='x', rotation=20)


# Menampilkan dashboard
plt.suptitle('DASHBOARD ANALISIS PENJUALAN NUSANTARA MART',
             fontsize=15, fontweight='bold')

plt.tight_layout()
plt.show()
```
